```python
#Load data
import pandas as pd
import numpy as np

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/paisabazar dataset.csv')
```

```python
#Inspect
print("Shape:", df.shape)
print("Unique customers:", df['Customer_ID'].nunique())
print("Months:", sorted(df['Month'].unique()))
print("Duplicates:", df.duplicated().sum())
```

```python
#Remove Duplicates aur sensitive columns
df = df.drop_duplicates()
df = df.drop(columns=['ID', 'Name', 'SSN'], errors='ignore')
```

```python
#Data types aur NM fix
df['Age'] = df['Age'].astype(int)
for c in ['Num_Bank_Accounts', 'Num_Credit_Card', 'Num_of_Loan',
          'Num_of_Delayed_Payment', 'Num_Credit_Inquiries', 'Delay_from_due_date']:
    df[c] = df[c].astype(int)

df['Payment_of_Min_Amount'] = df['Payment_of_Min_Amount'].replace('NM', 'Unknown')
```

```python
#New Columns
df['Age_Group'] = pd.cut(df['Age'], [0, 25, 35, 45, 100],
                         labels=['Up to 25', '26-35', '36-45', '46+'])
df['Income_Band'] = pd.qcut(df['Annual_Income'], 3, labels=['Low', 'Mid', 'High'])
df['Utilization_Band'] = pd.qcut(df['Credit_Utilization_Ratio'], 4,
                                 labels=['Low', 'Medium', 'High', 'Very High'])
df['Delay_Band'] = pd.cut(df['Delay_from_due_date'], [-np.inf, 0, 15, 30, np.inf],
                          labels=['On time', '1-15 days', '16-30 days', '30+ days'])
df['Debt_to_Income'] = np.where(df['Annual_Income'] > 0,
                                df['Outstanding_Debt'] / df['Annual_Income'], np.nan)
```

```python
#Check and Export
print(df.shape)
print(df['Credit_Score'].value_counts())
print(df.isnull().sum()[df.isnull().sum() > 0])

df.to_csv('paisabazaar_clean.csv', index=False)
from google.colab import files
files.download('paisabazaar_clean.csv')
```
